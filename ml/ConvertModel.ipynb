```python
import torch
import torch.nn as nn
from torchvision import models

# Step 1: Define the model architecture same as during training
model = models.resnet18(pretrained=False)
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 2)  # 4 classes (adjust if your classes are different)

# Step 2: Load the saved weights (.pth)
model.load_state_dict(torch.load("resnet18_leaf_nonleaf.pth", map_location=torch.device('cpu')))

# Step 3: Set model to evaluation mode
model.eval()

# Step 4: Create a dummy input tensor (same input size used in training)
example_input = torch.randn(1, 3, 224, 224)

# Step 5: Trace the model with the dummy input to create a TorchScript module
traced_script_module = torch.jit.trace(model, example_input)

# Step 6: Save the TorchScript module as a .pt file
traced_script_module.save("resnet18_leaf_nonleaf.pt")

print("Model has been successfully converted and saved as TorchScript (.pt) file!")
```
